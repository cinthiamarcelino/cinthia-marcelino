# Cenário - Gasolina

```python
df_media_consumo_ano_gasolina = df[(df.Produto=='GASOLINA')]
```

```python
df_media_consumo_ano_gasolina[0:5]
```

```python
df_media_consumo_ano_gasolina.reset_index(inplace=True)
```

```python
df_media_consumo_ano_gasolina[0:5]
```

```python
df_media_consumo_ano_gasolina['Ano'] = df_media_consumo_ano_gasolina['Ano'].astype(str)
df_media_consumo_ano_gasolina['Valor'] = round(df_media_consumo_ano_gasolina['Valor'],2).astype(float)
```

```python
df_media_consumo_ano_gasolina.info()
```

```python
df_media_consumo_ano_gasolina[0:5]
```

# Gráfico Valor Médio Preço Gasolina por ano

```python
fig, ax = plt.subplots(figsize=(10,5), facecolor='white', dpi= 80)
df_media_anos_gasolina_Y = df_media_consumo_ano_gasolina['Valor']
df_media_anos_gasolina_X = df_media_consumo_ano_gasolina['Ano']


plt.bar(df_media_anos_gasolina_X, df_media_anos_gasolina_Y,color='c',width=0.5)

plt.legend('')
plt.xlabel('Gasolina')
plt.ylabel('R$')
plt.title('Média Valor de Venda Gasolina por ano (2004 - 2021)')

# function to add value labels
def addlabels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i],y[i], ha = 'center')
addlabels(df_media_anos_gasolina_X, df_media_anos_gasolina_Y)
plt.show()
```
